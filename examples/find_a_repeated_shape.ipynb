## Locate the nearest match

Each window contains five samples. Row 0 describes the first five samples. Its `I_` entry gives the nearest neighbor’s start index and its `P_` entry gives their distance after z-normalization.

```python
import flashmp

signal = [1, 3, 2, 5, 4, 0, -2, 0, 8, 12, 10, 16, 14]
profile = flashmp.stump(signal, m=5)

match = int(profile.I_[0])
print(match, signal[match:match + 5])
print(f"{profile.P_[0]:.6f}")
```

The match starts at sample 8. Its five values equal twice the values in the first window plus six. The normalized distance rounds to zero.

```python
assert match == 8
assert profile.P_[0] < 1e-6
assert signal[match:match + 5] == [2 * x + 6 for x in signal[:5]]
```

## Try your own signal

Replace `signal` with numeric observations in time order and choose `m` to span the duration of the pattern you want to find. Divide a duration in seconds by the sampling interval in seconds to get a window length in samples.
